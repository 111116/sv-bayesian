```python
from PIL import Image

def scale_image(input_image_path, output_image_path, width=None, height=None):
    with Image.open(input_image_path) as image:
        w, h = image.size

        # If no specific width and height provided, we reduce the size by half
        if width is None and height is None:
            width = w // 2
            height = h // 2

        # If only one dimension is provided (either width or height), 
        # then maintain the aspect ratio of the image
        elif width is not None and height is None:
            height = int((width * h) / w)
        elif height is not None and width is None:
            width = int((height * w) / h)
        print("scaling to ", width, "x", height)

        scaled_image = image.resize((width, height))
        scaled_image.save(output_image_path)
```

```python
# Example usage
scale_image("wood_olive_wood_olive_basecolor.png", "wood_olive_wood_olive_basecolor_midres.png")
scale_image("wood_olive_wood_olive_roughness.png", "wood_olive_wood_olive_roughness_midres.png")
```
