### multivariate normal

```python
from scipy.stats import multivariate_normal
a = torch.ones((2))*2
multivariate_normal.pdf(a, a, a, allow_singular=True)
```

```python
from torch.distributions import MultivariateNormal
a = torch.ones((2)).cuda()*2
cov = torch.diag(a)
dist = MultivariateNormal(a, cov)
```

```python
torch.exp(dist.log_prob(a)).item()
```
