```python
res = 100
grid_x = np.linspace(0, 1, res)
grid_y = np.linspace(0, 1, res)
grid_X, grid_Y = np.meshgrid(grid_x, grid_y)
```

```python
import math

def gaussian(x, mu=0, sigma=1):
    """
    Computes the Gaussian function for given x, mean (mu) and standard deviation (sigma).
    This version is compatible with both scalars and numpy arrays.
    """
    coefficient = 1 / (sigma * np.sqrt(2 * np.pi))
    exponent = -((x - mu) ** 2) / (2 * sigma ** 2)
    
    return coefficient * np.exp(exponent)

```

```python
x = np.linspace(-4, 4, res)
y = [gaussian(t) for t in x]
plt.plot(x,y)
```

```python
def plot(f):
    # Create a grid of points over [0,1] x [0,1]
    x = np.linspace(0, 1, 100)
    y = np.linspace(0, 1, 100)
    X, Y = np.meshgrid(x, y)
    Z = f(X, Y)

    # Plot the function f(x, y)
    plt.figure(figsize=(8, 6))
    plt.contourf(X, Y, Z, 20, cmap="viridis")
    plt.colorbar()
    plt.title("f over [0,1] x [0,1]")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True)
    plt.show()
```

```python
def f(x,y):
    return gaussian(0.6*x+0.4*y, 0.5, 0.05) * gaussian(0.3*x+0.7*y, 0.6, 0.05)

plot(f)
```

```python
def f(x,y):
    return gaussian(0.5*np.exp(x) + 0.5*np.exp(y), 1.7, 0.1)

plot(f)
```
