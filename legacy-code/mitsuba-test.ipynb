```python
import times

class Timer:
    def __enter__(self):
        self.start_time = time.time()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.end_time = time.time()
        elapsed_time = self.end_time - self.start_time
        print(f"Time elapsed: {elapsed_time:.4f} seconds")
```

```python
scene = mi.load_file("assets/sphere_scene.xml")
img = mi.render(scene, spp=512)
plt.figure(figsize=(6,6))
plt.imshow(img**0.454)
```

```python
# performance benchmark
img = mi.render(scene, spp=128)
for _ in range(3):
    with Timer():
#         for _ in range(40):
        img = mi.render(scene, spp=128).torch()
```

```python
aov_integrator = mi.load_dict({
    'type': 'aov',
    'aovs': 'shape_index:shape_index,uv:uv,duv_dx:duv_dx,duv_dy:duv_dy'
})
```

```python
res = 128
valid = None
converge_counter = 0
for it in range(1000):
    # render aux buffers & extract
    auximg = mi.render(scene, spp=1, integrator=aov_integrator, seed=torch.randint(2147483647, (1,)).item())
    aux = auximg.torch()
    shapeid = aux[...,0]
    uv = aux[...,1:3]
    duvdx = aux[...,3:5]
    duvdy = aux[...,5:7]
    # check if each pixel is valid (at center of texel)
    uv_valid = (((uv*res).frac() > 0.3) & ((uv*res).frac() < 0.7)).all(dim=2)
    duv_valid = (((duvdx*res).abs() < 0.3) & ((duvdy*res).abs() < 0.3)).all(dim=2)
    if valid is None:
        valid = torch.ones_like(uv_valid)
    if (valid == valid & uv_valid & duv_valid).all():
        converge_counter += 1
    else:
        converge_counter = 0
    if converge_counter > 20:
        break
    valid &= uv_valid & duv_valid
print(it, "iterations elapsed")
```
